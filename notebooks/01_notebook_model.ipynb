```python
import sys; sys.path.append("..")
from src.data import build
from src.utils import evaluate, log_result
from sklearn.ensemble import RandomForestRegressor

d = build(variant="base", scaled=False)
model = RandomForestRegressor(n_estimators=200, max_depth=10, n_jobs=-1, random_state=0)
model.fit(d["X_train"], d["y_train"])
m = evaluate(d["y_val"], model.predict(d["X_val"]))
log_result("rf", "basec", "n=200,depth=10", "val", m)
```
